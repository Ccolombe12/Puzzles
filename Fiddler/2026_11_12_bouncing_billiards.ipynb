# Part 1:

> In the fictional game of “bouncing billiards,” a cue ball that has a radius of 1 is centered at the coordinates (0, 0) on a billiard table. There is also a red ball with radius 1, centered at (0, 5). The table has a top wall at y = 10 and two side walls at x = 5 and x = −5 that extend infinitely downward. The diagram below illustrates the setup.
>
> The goal of the game is to strike the cue ball with a stick and have the cue ball hit the red ball. The table and its walls are frictionless, so the balls don’t slow down as they roll.
> Now it’s your turn to play a round of bouncing billiards. Let’s say, since you’re new to the game, you have equal chances of hitting the cue ball in any direction.
>
> Your friend challenges you to make the cue ball hit the red ball before the cue ball makes contact with the walls _twice_. In other words, the cue ball can hit the red ball directly, or it can bounce off a wall exactly once before hitting the red ball.
>
> What’s the probability that you’ll succeed in your friend’s challenge?


# Extra Credit

> *(Paraphrased.)* Now suppose any number of bounces is allowed. What is the probability that the cue ball eventually hits the red ball?

# Solution

Let us first establish some notation. We say the cue ball $C$ is at position $(0, 0)$ in the plane. Let $w$ be the width of the hallway the balls lie in. We define $\theta$ to be the direction that $C$ will initially move in after being struck with $\theta = 0$ be pointing along the $y+$ direction and $\theta$ is increasing in the CW direction. Let $T_{0, 0}$ define the "target" ball at position $(x_0, y_0)$. (in our problem $x_0 = 0$ and $y_0$ is a given) Let the hallway extend exactly $w/2$ units past the center of $T_{0,0}$ in the $+y$ direction. We will expand on this choice of notation momentarily.

Often with problems that involve reflections, we can invoke some symmetry to gain some understanding. Here whenever the $C$ bounces off a wall, we can imagine the $C$ continuing in a straight line, passing through the wall into the region reflected about the axis of reflection.

Multiple bounces can be handled via multiple reflections in this manner. We can reflect the target ball $T_{0,0}$ into reflection space to create other "target balls" $T_{i, j}$ which represent the $T_{0, 0}$ reflected horizontally $i$ times and vertically $j$ times. We say that $T_{i, j}$ lies at position $(x_i, y_j)$. We draw out several of these reflected target balls in yellow below.

Note that $C$ may only bounce vertically one time before either hitting $T_{0,0}$ or missing and bouncing away forever: bounces off the side walls never change the sign of the vertical velocity, so once $C$ is heading down it stays heading down. Therefore, only one row of vertical bounce target are considered.



If WLOG we assume that the initial direction of the cue ball is to the right in some capacity, then if it were to move in a straight line (passing right through the walls) and hit either the target ball or one of the many reflected target balls, then we know it would have hit the target ball in reality. Note that in the event the first ball hit is at position $(x_i, y_j)$, then the number of bounces the ball took to hit the real target ball is $i + j$. Note that we say _first_ target ball since there may be the case that the cue ball could hit many target balls if it too could pass through them.

With this established, our goal is to determine the probability that one of these target balls is hit.

First we need to find an expression for $(x_i, y_j)$. The axis of reflection symmetry for the first horizontal collision is $x = w / 2 - r$ (the cue ball's *center* bounces when the ball's edge touches the wall), and consecutive axes are separated by $2(w/2 - r) = w - 2r$.  Similarly for the first (and only) vertical collision, the line of reflection is $y = y_0 + w/2 - r$. Using this, it is easy to show that $T_{i, j}$ would lie at 
$$
\begin{aligned}
x_i & = 2 i \left(\frac{w}{2} - r\right) \\
y_j & = 2 j \left(\frac{w}{2} - r\right) + y_0
\end{aligned}$$

It then follows that we can define the distance between $C$ and $T_{i, j}$ as $d_{i,j} \coloneqq \sqrt{x_i ^2 + y_j ^2}$.

Now for some $\theta$, $C$ will be sent on a direct collision course with $T_{i, j}$, since $C$ and the target have non-zero volumes there is some non-zero-width interval about $\theta$, $I_{i, j}$ such that for all $\theta' \in I_{i,j}$ $C$ will collide with $T_{i, j}$. The most obvious way to find this interval is to treat $\theta = \theta_{i, j}$ as the *direct* head-on-collision angle between the two balls and then find the width of the angle since, by symmetry, $\theta_{i, j}$ will be midpoint of $I_{i, j}$.

Recall that $\theta$ is measured from the $+y$ axis and we easily find that  $$\theta_{i, j} = \arctan(x_i/y_j)$$. Furthermore, we can use the diagram below to find the width of the interval, denoted $\phi_{i,j}$.




Some simple trigonometry yields that $$\phi_{i,j} = \arcsin(2 r / d_{i,j})$$

Therefore, we can now write the interval $I_{i, j}$ explicitly as 
$$I_{i, j} = [\theta_{i, j} -\phi_{i,j} ,\theta_{i, j} + \phi_{i,j} ] = [ \arctan(x_i/y_j)- \arcsin(2 r / d_{i,j}), \arctan(x_i/y_j)+ \arcsin(2 r / d_{i,j})]$$

One last observation before we are ready to compute the solution. Suppose we focus on just the $j = 0$ target balls. For some angles $\theta$ only one ball in row $j = 0$ is hit, for some none are hit, and for others *many* are hit. As $\theta \to \pi/2$, $C$ is almost certain to hit a target, since it bounces many times while making very little vertical progress. In fact there is a critical angle $\theta_*$ past which $C$ is *guaranteed* to hit a target ball in a given row. To find it, think about the gaps between neighboring targets $T_{i,j}$ and $T_{i+1,j}$: a direction misses both only if $C$ can slip *between* them, i.e. only if $I_{i,j}$ and $I_{i+1,j}$ don't overlap.

We observe that the gaps in each row eventually close, and that once a gap closes, every gap further out in that row stays closed. So for each row $j$ let $i^*_j$ be the first index with $I_{i^*_j, j} \cap I_{i^*_j + 1, j} \neq \emptyset$. Every direction in $[\theta_{i^*_j,j} - \phi_{i^*_j,j},\, \pi/2)$ then hits row $j$, and only the finitely many intervals before it need to be handled individually.

## Turning intervals into a probability

The initial direction is uniform on $[0, 2\pi)$. Any direction pointing downward ($\theta > \pi/2$) never hits anything, and the setup is symmetric about the $y$-axis, so it is enough to measure the union $U$ of the hit intervals on $[0, \pi/2]$:

$$
\Pr(\text{hit}) = \frac{2\,|U|}{2\pi} = \frac{|U|}{\pi}.
$$

**Part 1 (at most one bounce).** Target $T_{i,j}$ costs $i + j$ bounces, so the candidates are $T_{0,0}$, $T_{1,0}$ and $T_{0,1}$. But $T_{0,1}$ sits on the same ray as $T_{0,0}$, only farther away, so $I_{0,1} \subset I_{0,0}$: any shot aimed at the top-wall reflection hits the real target first. So

$$
U_1 = I_{0,0} \cup I_{1,0}.
$$

**Part 2 (any number of bounces).** Only rows $j = 0$ and $j = 1$ exist, and each is cut off at its critical index:

$$
U_2 = \bigcup_{j \in \{0, 1\}} \left( \bigcup_{i < i^*_j} I_{i,j} \;\cup\; \left[\theta_{i^*_j,j} - \phi_{i^*_j,j},\, \frac{\pi}{2}\right) \right).
$$

The code below builds these intervals row by row (stopping at $i^*_j$), merges overlapping ones, and adds up their lengths.


```python
# generate intervals
from math import atan2, sqrt, pi, asin
import numpy as np
import matplotlib.pyplot as plt

INF = 10**20


def get_next_reflection(row, k, w, r, y0):
    xk = 2 * k * (w / 2 - r)
    yk = 2 * row * (w / 2 - r) + y0
    return xk, yk


def compute_interval(r, x, y):
    d = sqrt(x**2 + y**2)
    theta = atan2(x, y)
    phi = asin(2 * r / d)
    interval = [
        max(theta - phi, 0),
        theta + phi,
    ]  # make sure it positive so we can leverage symmtery
    return interval


def compute_probability(r, w, y0):
    # get the part1 and part2 solutions for balls of radius r,
    # a hallway of width w (and space behind the ball of width w / 2),
    # and an initial distance of y0 away.

    def build_intervals(row, max_bounces=INF):
        # for a given row (0 being the row equal to target) of reflected
        # red balls, compute the angle intervals that hit them.
        intervals = []
        k = 0
        while k <= max_bounces:
            x, y = get_next_reflection(row, k, w, r, y0)
            lo, hi = compute_interval(r, x, y)
            if not intervals:
                intervals.append([lo, hi])
                k += 1
                continue

            prev_hi = intervals[-1][-1]
            if lo <= prev_hi and max_bounces == INF:  # only for infinite bounce case.
                plo, hi = intervals.pop()
                intervals.append([plo, pi / 2])
                break
            intervals.append([lo, hi])
            k += 1
        return intervals

    def merge_intervals(intervals):
        intervals.sort()
        intervals.append([INF, INF])
        merged = []
        lo, hi = None, None
        for s, e in intervals:
            if lo is None:
                lo, hi = s, e
            elif s > hi:
                merged.append([lo, hi])
                lo, hi = s, e
            else:
                hi = max(e, hi)
        return merged

    # part 1
    my_intervals = []
    for row in range(1):
        my_intervals.extend(build_intervals(row, max_bounces=1))
    merged1 = merge_intervals(my_intervals)
    part1 = sum(hi - lo for lo, hi in merged1) / pi

    # part 2
    my_intervals = []
    for row in range(2):
        my_intervals.extend(build_intervals(row))
    merged2 = merge_intervals(my_intervals)
    part2 = sum(hi - lo for lo, hi in merged2) / pi
    return part1, part2, merged1, merged2


# Compute the answer.

W = 10  # hallway width
R = 1
Y0 = 5

part1, part2, m1, m2 = compute_probability(R, W, Y0)
print(
    f"Probability of hitting the red ball with at most one bounce: {part1:.4f}\n"
    f"Probability of hitting the red ball with any number of bounces: {part2:.4f}"
)
```

# Answer

With $r = 1$, $w = 10$ and $y_0 = 5$, the code above gives

$$
\Pr\left[\text{hit with at most one bounce}\right] \approx \boxed{0.2670}, \qquad \Pr\left[\text{hit with any number of bounces}\right]\approx \boxed{0.4628}.
$$

# Visualizing the hit directions
For fun, let's visualize the intervals at which $C$ would hit a target!

Each panel shades the directions $\theta$ that hit the red ball, drawn on the upper unit semicircle with $\theta = 0$ straight up. The left panel allows at most one bounce and the right allows any number of bounces.

```python
def plot_hit_directions(intervals, ax=None, color="tab:blue", title=None):
    """Shade the union of [lo, hi] angle intervals on the top unit semicircle.
    Angles are measured from the +y axis, clockwise; each interval is
    mirrored across the y-axis."""
    if ax is None:
        _, ax = plt.subplots(figsize=(6, 3.4))
    t = np.linspace(-np.pi / 2, np.pi / 2, 400)
    ax.plot(np.sin(t), np.cos(t), "k", lw=2.5)  
    ax.plot([-1, 1], [0, 0], "k", lw=2.5)  
    for lo, hi in intervals:
        t = np.linspace(lo, hi, max(2, int(200 * (hi - lo)) + 2))
        for s in (1, -1):  
            xs = np.r_[0, s * np.sin(t), 0]
            ys = np.r_[0, np.cos(t), 0]
            ax.fill(xs, ys, color=color, alpha=0.5, lw=0)
    ax.text(0, 1.04, r"$\theta = 0$", ha="center", va="bottom")
    ax.text(1.03, 0, r"$\frac{\pi}{2}$", ha="left", va="center")
    ax.text(-1.03, 0, r"$-\frac{\pi}{2}$", ha="right", va="center")
    ax.set_aspect("equal")
    ax.axis("off")
    if title:
        ax.set_title(title, pad=18)
    return ax


fig, axes = plt.subplots(1, 2, figsize=(12, 3.6))
plot_hit_directions(m1, ax=axes[0], title=r"Cue ball collision angles, $\leq$ 1 bounce")
plot_hit_directions(m2, ax=axes[1], title="Cue ball collision angles, any # bounces")
```

## Varying the hallway width

Holding $r = 1$ and $y_0 = 5$ fixed, the animation below sweeps $w$ from $2.5$ to $60$. Note that in this parameterization the top wall sits $w/2$ past the red ball, so widening the hallway also pushes the top wall back.

```python
import numpy as np
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import Image

R, Y0 = 1, 5
Ws = np.linspace(2.5, 60, 80)  # need W/2 > R so the ball fits

fig, axes = plt.subplots(1, 2, figsize=(12, 4))


def draw(W):
    p1, p2, m1, m2 = compute_probability(R, W, Y0)
    for ax in axes:
        ax.clear()
    plot_hit_directions(m1, ax=axes[0], title=rf"≤ 1 bounce: Pr = {p1:.3f}")
    plot_hit_directions(m2, ax=axes[1], title=f"any bounces: Pr = {p2:.3f}")
    fig.suptitle(rf"$w$ = {W:.2f}")


anim = FuncAnimation(fig, draw, frames=Ws)
anim.save("billiards_W.gif", writer=PillowWriter(fps=7))
plt.close(fig)
Image("billiards_W.gif")
```
