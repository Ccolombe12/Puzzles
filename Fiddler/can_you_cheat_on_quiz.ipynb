# Part 1


WLOG let the initial correct answer for question $0$ be $A$ (which we denote as index 1). For question $k$ we want to know the probability that answer $i$ is correct, which we denote as $P_i^k$

Initially, we have that $\mathbf{P^0} = (1, 0, 0, 0)$. For $k = 2$ it is easy to show $\mathbf{P^1} = (0, 1/3, 1/3, 1/3)$.

In general we have that 

$$
\begin{equation}
\begin{aligned}
P^k_i & = (1 - P^{k-1}_i)\cdot \frac{1}{3} + P^{k -1} \cdot 0 \\ 
& = (1 - P^{k-1}_i)\cdot \frac{1}{3}
\end{aligned}
\end{equation}
$$

where we condition on the probability that $i$ was the answer to the previous problem.

We can solve this linear reccurence relation with base case $P^0_1 = 1$ and $P_i^0 = 0$ for $i \neq 1$.

We find that 

$$
\begin{equation}
P_i^k  = 
    \begin{cases}
    \frac{1}{4}\left(1 -(\frac{-1}{3})^{k - 1}\right) & i = 1\\
    \frac{1}{4}\left(1 -(\frac{-1}{3})^k \right) & i \neq 1
    \end{cases}
\end{equation}
$$

for $k \geq 0$. One nice observation here is that $P_{i}^k \to 1/4$ as $k\to \infty$, which implies that the distribution does eventually revert to the uniform distribution which is what we would expect.

Now, by the linearity of expectation, if we seek to maximize the expected score then at each turn, we should choose the maximal element of $\mathbf{P^k}$.

In general for a quiz with $K + 1$ questions the expected score, given the first answer is known 

$$\frac{1}{4}\sum_{k = 0}^K \max\left(\left(1 -(\frac{-1}{3})^k\right), \left(1 -(\frac{-1}{3})^{k - 1}\right) \right)$$

Exploring with a few small values of $k$, we see the exponents that get chosen are $-1, 1, 1, 3, 3, \ldots$. We can simplify our summation and then find a closed form in mathematica for $K + 1$ questions is 

$$\boxed{S(K + 1) =  \frac{K}{4} + \frac{19}{16} + \frac{(-1)^{K}-4}{16\cdot 3^{K}}}$$

Which implies that for $K + 1 = 3$ questions, the maximal expected score is $\boxed{S(3) = 5/3\approx 1.66}$ which is $1$ point from the first problem $1/3$ from the second and then $1/3$ from the last. We actually guess $A$ again on the last question! This is the pattern, we guess $A$ on even question indices and not $A$ otherwise. The grade is a $55$ out of $100$ which is better than the $25$ he would have expected to get before! 





# Part 2

Now $K + 1 = 7$ and we need to pick which question we reveal. Say we pick question $k$ where $k$ is not the first or last question, then we essentially have split the problem into two separate problems. The distribution of correct answers propogates forwards and backwards from $k$ identically as before. Thus the best score we can get if we pick question $k$, denoted $T(k)$ is: 

$$
\begin{equation}
T(k) := S(k + 1) + S(K - k + 1) - 1
\end{equation} $$ 

where we need to subtract one since these both assume we get the free point from cheating and it is double counted. 

If $k$ is an endpoint, then $T(0) = $S(K + 1)$, the previous answer. 

We need to choose $k$ to maximize the score. It seems intuitive that the symmetric answer would be maximizing. As we get further away from a known answer (in either) direction, the distributions absolute deviation from the uniform decreases. We want both sides to be as close to the known answer as possible. Thus $k^* = 3$ the middle problem is the one to cheat on. 


Therefore $$\boxed{2 S(3) - 1 = \frac{77}{27} \approx 2.85}$$. This equates to roughly a $40/100$. If he had instead cheated on the first problem, he would have gotten a $38/100$ in expectation.
