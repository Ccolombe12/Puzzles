# Part 1

> Inspired by Delta Airlines, you and a graphic designer colleague are crafting a new logo for Fiddler Airlines. Your colleague starts with an equilateral triangle with a side length of 1 unit. Next, you pick a random point inside the triangle, giving you four points: the three corners of the triangle and this fourth point inside. There are several ([simple](https://en.wikipedia.org/wiki/Simple_polygon)) quadrilaterals that can be drawn whose four vertices are these four points. Your colleague picks one of these quadrilaterals at random, with each having an equal probability. This quadrilateral will be the new logo.
>
> On average, what can you expect the area of the quadrilateral to be?

# Solution

Let the area of the entire triangle $T$ be given by the constant $a = \sqrt{3}/4$. When we place a uniformly random point $P$ in the triangle, we can form 3 potential quadrilaterals. If we connect the point $P$ with each of the vertices of $T$, then we form 3 disjoint smaller triangles that partition $T$. Each potential quadrilateral is formed by removing exactly one of these triangles. Let the areas of the triangles be the random variables $A_1$, $A_2$, $A_3$. Then the expected area of our quadrilateral, conditioning on which triangle is removed, can be written as 

$$
\begin{aligned}
Q & = \frac{1}{3}\mathbb{E}[a - A_1] + \frac{1}{3}\mathbb{E}[a - A_2] + \frac{1}{3}\mathbb{E}[a - A_3]\\ 
 & = a - \frac{1}{3}\mathbb{E}[A_1 + A_2 + A_3] \\
 & = a - \frac{1}{3}\mathbb{E}[a]\\ 
 & = \frac{2}{3}a \\ 
 & = \boxed{ \frac{1}{2\sqrt{3}}}.
\end{aligned}
$$

Thus the area of the quadrilateral formed will be $2/3$ of the area of $T$ on average. 

# Extra Credit

> While picking random points inside the equilateral triangle with side length 1, you and your colleague notice that some points are “special.” When one of these special points is connected with the triangle’s three vertices, it creates three smaller triangles whose areas form an arithmetic sequence.
>
> After further investigation, you realize these special points all belong to a locus, the total length of which is $L$. What is this length $L$?

# Solution

Suppose WLOG that the smallest triangle has area $A_1$. Then, to form an arithmetic sequence, the other areas must be $A_2 = A_1 + d$ and $A_3 = A_1 + 2d$ for some non-negative $d$. If we add the areas, we find

$$
A_1 + A_2 + A_3 = 3(A_1 + d) = a \implies \boxed{A_1 + d = \frac{1}{3}a}
$$

In plain English, this implies that the *second-largest triangle* will make up $1/3$ of the total area of $T$. We can now use some analytic geometry. Consider the figure below:



We anchor the lower-left vertex of $T$ to $(0, 0)$ and axis-align the lower edge. For ease of computation, we will consider the case where the triangle formed by the lower two vertices is $A_2$, the right two $A_1$, and the left two $A_3$. 
For this to be the case, $P$ must fall in the region $R$ defined by the following constraints:

$$
R = \left\{(x, y) \in \mathbb{R}^2 \bigg| \; y \leq \frac{1}{\sqrt{3}}\; x, \quad \; y \geq - \frac{1}{\sqrt{3}} \;(x - 1), \quad y \leq - \sqrt{3} \;(x - 1)\right\}
$$



For the area of the lower triangle $A_2$ to be equal to $\frac{1}{3}a$, we require that

$$
\frac{1\cdot y}{2} = \frac{1}{3}a = \frac{1}{4\sqrt{3}} \implies \boxed{y = \frac{1}{2\sqrt{3}}}
$$

We also require that $A_3 - A_2 = A_2 - A_1$, but a little algebra shows this reduces to the condition $A_1 + A_3 = \frac{2}{3} a$, which always holds when $A_2 = \frac{1}{3}a$. So we have a segment $\ell$ in $R$ that produces triangles with the desired properties. One thing to note is that this segment is at the same $y$-coordinate as the centroid of $T$. Indeed, if we evaluate the angle bisector of the lower-left vertex (the blue line in the figure), we see it has equation
$$
y = \frac{1}{\sqrt{3}}x,
$$
which, when evaluated at $x = 1/2$, yields $y = \frac{1}{2\sqrt{3}}$. We now want to find the length of $\ell$ in $R$.

By solving the equation $\frac{1}{2\sqrt{3}} = - \sqrt{3} (x_r - 1)$, we find that $\ell$ intersects the right edge of $T$ at $x_r = 5/6$. Subtracting its start at $x = 1/2$, we find that $\ell$ has length $\frac{1}{3}$. 
By symmetry (one copy for each ordering of the areas of the sub-triangles), we see there are exactly 6 copies of $\ell$, which implies their total length is $$\boxed{L = 2}.$$
