> A surfer is riding along a crest of a sinusoidal wave given by the function $y = 2\sin(x−t)$, where y is the height above sea level in meters, x is horizontal position in meters, and t is time in seconds. Based on this, the surfer knows that her speed—and the speed of the wave—is 1 meter per second.
>
> But hold on—it appears that another wave is approaching from the opposite direction. This wave has half the amplitude but twice the speed: $y = sin(x+2t)$.
>
> The two waves, the forward and the backward, ”interfere,” meaning they combine. In other words, the height of the water at position x and time t is given by $y = 2\sin(x−t) + \sin(x+2t)$. Nevertheless, the surfer stays atop a crest of the combined wave.
> 
> As the wave moves along, what are the surfer’s slowest and fastest speeds, in meters per second? (Edit: My intent was to ask for speed in the horizontal direction. If you include horizontal and vertical components, I’ll still mark it as correct.)

# Solution

We can start by getting a sense of what the combined wave looks like. 



Here the blue wave corresponds to $y_1(x, t) = 2\sin(x - t)$, orange to $y_2(x, t) = \sin(x+2t)$, and green to the combined wave $y_3 = 2\sin(x−t) + \sin(x+2t)$.
We can see some interesting movement from the green wave as it seems to freeze in place and then drop down in a jerky motion. We need to track the movement of a single peak of the combined wave through time as that is where our surfer will sit. 


Before we do any work in tracking the surfer's path through time, the wave is a bit cumbersome to work with as is. We can use the substitution $\phi = x - t$ which gives us 

$$ 
y_3(\phi, t) = 2 \sin \phi + \sin (\phi + 3t).
$$ 
We can now use the angle-sum identity for sin to write 

$$
y_3(\phi, t) = 2 \sin \phi + \sin \phi \cos 3t  + \sin 3t \cos \phi 
$$
To find the location of the peak we are interested in we can take the derivative of the combined wave in $x$ and solve for the maximum, treating $t$ as a constant. 

We find 
$$
\frac{dy_3(\phi(x,t), t)}{d x} = 2 \cos \phi + \cos \phi \cos 3t  - \sin 3t \sin \phi 
$$
Which, upon setting equal to zero and isolating for $\phi$ gives 

$$
2 \cos \phi + \cos \phi \cos 3t  - \sin 3t \sin \phi = 0 \implies \tan \phi = \frac{2 + \cos 3t}{\sin 3t} \implies \boxed{x^*(t) = t + \arctan\left(\frac{2 + \cos 3t}{\sin 3t}\right) + k \pi }.
$$

To find the particular extrema we are looking for, we note that the maximizer when $t = 0$ is $\pi/2$, which is what $x^*(t)$ with $k = 0$ gives us in the limit $t \to 0$, so we know $k = 0$ corresponds to the peak we are interested in. 

Now $x^*(t)$ is the location of our surfer on the combined wave over time. We can verify it by plotting the location of the surfer using our equation below.



This seems to confirm we have nailed the surfer's position. 

To find the velocity of the surfer as a function of time, we simply take $\frac{d x^*(t)}{d t}$ and using Mathematica, find 

$$
v^*(t) = \frac{4 \sin ^2\left(\frac{3 t}{2}\right)}{4 \cos (3 t)+5} 
$$
which we plot below.



Finding the minimum and maximum of this function with calculus or in Mathematica yields a minimum speed of $\boxed{v_{min} = 0}$ (at $t = 0$) and maximum of $\boxed{v_{max} = 4}$ (at the valley, $t = \pi$)




> The surfer rides the crest of the combined wave, $y_3(x, t) = 2\sin(x−t) + \sin(x+2t)$, for a very long time.

> What is her average speed, in meters per second? (Edit: My intent was to ask for speed in the horizontal direction. If you include horizontal and vertical components, I’ll still mark it as correct.)

# Solution
We note that the surfer's wave repeats periodically, so it is sufficient to analyze it in its first period. The two waves, initially together at $t = 0$, first recombine to the right at $t_f$.
We can solve for this by solving for the peak location of each wave (as we did above for the combined wave). We find $x_1^*(t) = t + \pi/2$ and $x_2^*(t) = \pi/2 -2t + 2\pi$.

Solving for the first $t > 0$ where these agree, we find $t_f = 2\pi/3$. Looking at wave 1, we see that it is at position $x_1^*(t_f) = \frac{7 \pi}{6}$ and this is where the waves recombine and thus where the period of our surfer's oscillation restarts.

Overall, per oscillation, our surfer moves a distance of $\Delta x = \frac{7 \pi}{6} - \pi/2 = \frac{2 \pi}{3}$ and does this over $\Delta t = t_f - 0 = \frac{2 \pi}{3}$ units of time, for an average horizontal speed (m/s) of 
$$
\boxed{\overline{v} = 1 }
$$
